## 0. Siapkan data (persiapan Sesi 5)

```python
import pandas as pd

df = pd.read_csv("ecommerce_transactions.csv")
df = df.drop_duplicates()
df = df[df["price"] > 0].copy()
df["city"] = df["city"].str.strip().str.title()
df["product"] = df["product"].str.strip().str.title()

print(len(df), df["payment_method"].notna().sum(), df["city"].notna().sum())
assert (len(df), df["payment_method"].notna().sum(), df["city"].notna().sum()) == (487, 457, 455)
```

Nilai kosong **tidak** diisi pada notebook ini. Baris tanpa `city` atau `payment_method` tidak masuk tabel pivot.

## Contoh: category × payment_method

```python
contoh = df.pivot_table(index="category", columns="payment_method",
                        values="order_id", aggfunc="count", fill_value=0)
print(contoh)
assert contoh.shape == (5, 4) and contoh.values.sum() == 457
```
